## Quick Look at the Data

The starter dataset provides evidence that the Refresh / Content Opportunity Scoring lane is worth exploring.

First, the correlation between search volume and impressions is approximately **0.001**, showing that search demand alone does not explain page visibility.

Second, the median age of declining pages is **216 days**, suggesting that older content may be more likely to require review, although age alone does not determine performance.

Finally, CTR differs across search position tiers. For example, keyword articles on page 1 have an average CTR of about **0.346**, while keyword articles in deeper positions average about **0.056** CTR. This indicates that ranking position is an important signal when evaluating content opportunities.

These observations suggest that combining several observable signals may produce a better ranking of pages for review than relying on a single metric.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

visible = df[df["impressions_90d"] > 0]

print("Number of pages:", len(df))

corr = visible["search_volume"].corr(visible["impressions_90d"])
print("Correlation between search volume and impressions:", round(corr, 3))

print("\nMedian content age by trend direction")
print(df.groupby("trend_direction")["content_age_days"].median())

print("\nAverage CTR by position tier and content type")
print(
    visible.groupby(["position_tier", "content_type"])["ctr"]
    .mean()
    .round(3)
)
```
